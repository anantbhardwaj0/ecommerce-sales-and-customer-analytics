### 3. Normalize City Names

Geolocation city names contain inconsistent accent usage and special
characters. For example, `são paulo` and `sao paulo` represent the same
city but are stored differently.

To standardize the categorical data, accents will be removed using
Unicode normalization. This will make equivalent city names consistent
without manually mapping individual cities.

Values such as `sp` and encoding-corrupted values will not be manually
replaced at this stage because they require separate validation.

```python
import unicodedata

def normalize_city(value):
    if pd.isna(value):
        return value

    value = str(value)

    # Remove accents
    value = ''.join(
        char for char in unicodedata.normalize('NFKD', value)
        if not unicodedata.combining(char)
    )

    return value

geolocation['geolocation_city'] = (
    geolocation['geolocation_city']
    .apply(normalize_city)
)

print("City names normalized successfully.")
```

## Step 9: Date & Time Validation

Date and time fields are validated to ensure that the order timeline is
logically consistent.

The expected sequence for an order is:

Purchase → Approval → Carrier Handover → Customer Delivery

Date relationships will be checked for impossible sequences without
modifying any records.

```python
print("=" * 60)
print("DATE & TIME CONSISTENCY CHECK")
print("=" * 60)

# Purchase should not occur after approval
purchase_after_approval = (
    orders['order_approved_at'].notna() &
    orders['order_purchase_timestamp'].notna() &
    (orders['order_purchase_timestamp'] > orders['order_approved_at'])
)

# Approval should not occur after carrier handover
approval_after_carrier = (
    orders['order_approved_at'].notna() &
    orders['order_delivered_carrier_date'].notna() &
    (orders['order_approved_at'] > orders['order_delivered_carrier_date'])
)

# Carrier handover should not occur after customer delivery
carrier_after_delivery = (
    orders['order_delivered_carrier_date'].notna() &
    orders['order_delivered_customer_date'].notna() &
    (orders['order_delivered_carrier_date'] >
     orders['order_delivered_customer_date'])
)

# Purchase should not occur after customer delivery
purchase_after_delivery = (
    orders['order_purchase_timestamp'].notna() &
    orders['order_delivered_customer_date'].notna() &
    (orders['order_purchase_timestamp'] >
     orders['order_delivered_customer_date'])
)

print("\nPurchase after approval:", purchase_after_approval.sum())
print("Approval after carrier:", approval_after_carrier.sum())
print("Carrier after delivery:", carrier_after_delivery.sum())
print("Purchase after delivery:", purchase_after_delivery.sum())
```

## 14. Delivery Delay by Seller State

Delivery delay is analyzed across seller states to identify geographical differences in delivery performance.

This analysis helps identify states with higher delivery delays and highlights potential areas for logistics and operational improvement.


```python
# Create seller-state delivery dataset

seller_delivery = delivered_orders.merge(
    order_items_clean[['order_id', 'seller_id']],
    on='order_id',
    how='inner'
).merge(
    sellers_clean[['seller_id', 'seller_state']],
    on='seller_id',
    how='left'
)

seller_delivery.head()
```

## 14.1 — Average Delivery Delay by Seller State

```python
# Average delivery delay by seller state

state_delivery_delay = seller_delivery.groupby('seller_state')['delivery_delay_days'].agg(
    Average_Delay_Days='mean',
    Number_of_Orders='count'
).reset_index()

state_delivery_delay = state_delivery_delay.sort_values(
    'Average_Delay_Days',
    ascending=False
)

state_delivery_delay
```

```python
# Late delivery rate by seller state

state_late_rate = seller_delivery.groupby('seller_state').agg(
    Total_Orders=('order_id', 'count'),
    Late_Orders=('delivery_performance', lambda x: (x == 'Late').sum())
).reset_index()

state_late_rate['Late_Delivery_Rate'] = (
    state_late_rate['Late_Orders'] /
    state_late_rate['Total_Orders'] * 100
)

state_late_rate = state_late_rate.sort_values(
    'Late_Delivery_Rate',
    ascending=False
)

state_late_rate
```

## 14.2 — Visualization

```python

#  Filter states with at least 100 orders

state_late_rate_filtered = state_late_rate[
    state_late_rate['Total_Orders'] >= 100
].copy()

state_late_rate_filtered
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

plt.bar(
    state_late_rate_filtered['seller_state'],
    state_late_rate_filtered['Late_Delivery_Rate']
)

plt.title('Late Delivery Rate by Seller State')
plt.xlabel('Seller State')
plt.ylabel('Late Delivery Rate (%)')
plt.xticks(rotation=45)

plt.show()
```
